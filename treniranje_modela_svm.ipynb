# **Importanje biblioteka**

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from sklearn.model_selection import GridSearchCV
```

**Ucitavanje datoteke**

```python
data = pd.read_csv("C:/Users/Jerons/Desktop/merged_features.csv")
print(data.head())
print("Broj uzoraka za svaku klasu")
print(data['event'].value_counts())
```

**Normalizacija**

```python
data_to_standardize=data.drop(columns=['event'])
scaler = StandardScaler()
standardized_data = scaler.fit_transform(data_to_standardize)
print("Standardizirani podaci")
print(standardized_data)
events_per_column=data['event']
```

**mapiranje klasi**

```python
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(events_per_column)

print("\nMapiranje klasa:")
for i, class_name in enumerate(label_encoder.classes_):
    print(f"{class_name}: {i}")
```

**odvajanje training i test**

```python
X_train, X_test, y_train, y_test = train_test_split(standardized_data, y_encoded, test_size=0.2, random_state=1)
print("\nDimenzije podataka nakon podjele:")
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

**treniranje modela**

```python
model = SVC(kernel='rbf', C=1.0, gamma='scale', decision_function_shape='ovr')
model.fit(X_train, y_train)
```

**provjera tocnosti modela**

```python
y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.2f}")
```

**pokusavanje podizanje tocnosti modela mijenjanjem parametara**

prvo gledamo najbolji kernel

```python
kernels = ['linear', 'rbf', 'poly']
for kernelnu in kernels:
    model= SVC(kernel=kernelnu, C=1.0, gamma='scale', decision_function_shape='ovr')
    model.fit(X_train,y_train)
    y_pred = model.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    print(f"Accuracy: {accuracy:.2f}")
    print(" for kernel "+ kernelnu)
    
```

gledamo koji ce biti najbolji Gamma

```python
gammas = [0.1, 1, 10, 100]
for gammanu in gammas:
    model= SVC(kernel='rbf', C=1.0, gamma=gammanu, decision_function_shape='ovr')
    model.fit(X_train,y_train)
    y_pred = model.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    print(f"Accuracy: {accuracy:.2f}")
    print(" for gamma "+ str(gammanu))
```

gledamo parametar c

```python
cset = [0.1, 1, 10, 100, 1000]
for csetnu in cset:
    model= SVC(kernel='rbf', C=csetnu, gamma='scale', decision_function_shape='ovr')
    model.fit(X_train,y_train)
    y_pred = model.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    print(f"Accuracy: {accuracy:.2f}")
    print(" for C "+ str(csetnu))
```

gledamo parametar degree

```python
degrees = [0, 1, 2, 3, 4, 5, 6]
for degreenu in degrees:
    model= SVC(kernel='rbf', C=1000, gamma='scale', decision_function_shape='ovr',degree=degreenu)
    model.fit(X_train,y_train)
    y_pred = model.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    print(f"Accuracy: {accuracy:.2f}")
    print(" for degree "+ str(degreenu))
```

gledamo decision function shape

```python
model= SVC(kernel='rbf', C=1000, gamma='scale', decision_function_shape='ovo')
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.2f}")
```

koristimo grid search da potvrdimo najbolje parametre

```python
degreeG = [0, 3]
cSetG = [100,1000,1500]
model = SVC(kernel='rbf')
gridSearch = GridSearchCV(estimator=model,param_grid={'degree':degreeG,'C':cSetG},cv=4)
gridSearch.fit(X_train,y_train)
print(f'The best parameter is {gridSearch.best_params_}')
print(f'The best score is {gridSearch.best_score_}')
print(f'The best estimator is {gridSearch.best_estimator_}')
```
